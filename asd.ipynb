# **Basic of LL**
---

```python
class Node:
    def __init__(self, data: int | None = None, next: Node | None = None) -> None:
        self.data = data
        self.next = next
```

```python
head = Node(1, None)
second = Node(2, None)
third = Node(3, None)
fourth = Node(4, None)

head.next = second
second.next = third
third.next = fourth
```

```python
temp = head
while temp != None:
    print(temp.data, end=" -> ")
    temp = temp.next
print(None)
```
